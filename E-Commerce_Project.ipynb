## Q 5. Find the Total Revenue generated by the E-Commerce company over all orders placed. (6 marks)

#### Hint: Calculate revenue of E-commerce company using the conditions mentioned above and then do the total (sum) of all to get total Revenue. 

### Ans 5

```python
ECom_Data
```

```python
ECom_Data.columns
```

```python
revenue_list = []

for price in ECom_Data['discounted_price']:
    if price > 600:
        revenue_list.append(price * 0.25)
    elif price > 350:
        revenue_list.append(price * 0.15)
    elif price > 100:
        revenue_list.append(price * 0.10)
    else:
        revenue_list.append(price * 0.05)

ECom_Data['ecommerce_revenue'] = revenue_list


#The revenue for the e-commerce company was calculated for each order based on predefined commission slabs applied to the discounted price. 
```

```python
ECom_Data.columns
```

```python
ECom_Data.head()
```

```python
total_revenue = ECom_Data['ecommerce_revenue'].sum()
total_revenue
#The total revenue was obtained by summing the revenue generated across all orders.
```

## Q6.Calculate the total BrandRevenue and list the top 10 Brand having maximum revenue in descending order (6 marks)
#### Hint: Total BrandRevenue is BrandRevenue which is generated after all the deductions. Also, BrandRevenue is different from the E-Commerce company's revenue.
#### Brand Revenue is the discounted price minus the Revenue (generated using the conditions given above Q5)


### Ans 6

```python
brand_revenue_list = []
for i in range(len(ECom_Data)):
    price = ECom_Data['discounted_price'][i]
    revenue = ECom_Data['ecommerce_revenue'][i]
    brand_revenue_list.append(price - revenue)
```

```python
ECom_Data['brand_revenue'] = brand_revenue_list
```

```python
#checking whether the column is added or not
ECom_Data.head()
```

```python
brand_total_revenue = (ECom_Data.groupby('Brand')['brand_revenue'].sum())
```

```python
ECom_Data['Year'] = ECom_Data['Order_Date'].dt.year
ECom_Data['Month'] = ECom_Data['Order_Date'].dt.month
```

```python
ECom_Data.head()

# now we create seperated columns for year and month
```

```python
region_monthly_revenue = (ECom_Data.groupby(['Region', 'Year', 'Month'])['ecommerce_revenue'].sum().reset_index())
region_monthly_revenue
```

```python
# To show Month + Year together, creating a simple label.
```

```python
region_monthly_revenue['Month_Year'] = (region_monthly_revenue['Month'].astype(str)+ '-' +region_monthly_revenue['Year'].astype(str))
region_monthly_revenue['Month_Year']
```

```python
regions = region_monthly_revenue['Region'].unique()
regions
```

```python
sns.lineplot(
    data=region_monthly_revenue,
    x='Month_Year',
    y='ecommerce_revenue',
    hue='Region',
    marker='o'
)

plt.title('Monthly E-Commerce Revenue by Region')
plt.xlabel('Month-Year')
plt.ylabel('Revenue')
plt.xticks(rotation=45)
plt.show()
```
